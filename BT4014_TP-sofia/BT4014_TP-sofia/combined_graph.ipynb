```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
epsilongraph = pd.read_csv("evaluations/evaluation_df.csv")
epsilongraph.head()
```

```python
annealingsm = pd.read_csv("evaluations/annealingSM_offline.csv")
annealingsm.head()
```

```python
lineargraph= pd.read_csv("evaluations/linearucb.csv",index_col=0)
lineargraph.head()
```

```python
thompson = pd.read_csv()
thompson.head()
```

```python
epsilongraph['algorithm'] = 'Epsilon Decay'
annealingsm['algorithm'] = 'Annealing SoftMax'
lineargraph['algorithm'] = 'Linear UCB'
thompson['algorithm'] = 'Contextual Thompson Sampling'
```

```python
combined_df = pd.concat([epsilongraph, annealingsm, lineargraph,thompson])

plt.figure(figsize=(10, 6))
for algorithm in combined_df['algorithm'].unique():
    subset = combined_df[combined_df['algorithm'] == algorithm]
    plt.plot(subset['time_step'], subset['cumulative_rewards'], label=algorithm)

# Customize the plot
plt.xlabel('Time Step')
plt.ylabel('Cumulative Rewards')
plt.title('Average Cumulative Rewards Comparison of Algorithms')
plt.legend()
plt.ylim(bottom=0)
plt.grid(True)
plt.show()
plt.savefig()

```
